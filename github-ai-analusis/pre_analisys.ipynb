## 0. Technical review of raw data

```python
raw_df = pd.read_csv('data/raw/github_ai_repositories.csv')
raw_df.head(1)
```

```python
# raw_df.drop_duplicates()
```

## 0.1 Remove url's cols

```python
raw_df2 = raw_df.loc[:, ~raw_df.columns.str.contains("url", case=False)]
raw_df2.head(1)
```

```python
raw_df2.info(verbose=True, max_cols=None, show_counts=True)
```

## 0.2 Preserving selected columns for analysis

```python
columns_to_keep = [
    "id",
    "name",
    "full_name",
    "description",
    "created_at",
    "updated_at",
    "pushed_at",
    "homepage",
    "size",
    "stargazers_count",
    "forks_count",
    "open_issues_count",
    "language",
    "topics",
    "owner.login",
    "owner.type",
    "license.name",
    "license.spdx_id",
    "fork",
    "archived",
    "has_issues",
    "has_wiki",
    "has_pages",
    "has_discussions",
]

df_cut = raw_df2[columns_to_keep].copy()
```

```python
len(columns_to_keep)
```

```python
df_cut.info()
```

## 0.3 Removing duplicates

```python
df_cut = df_cut.drop_duplicates()
# df_cut.head(1)
```

```python
df_cut.duplicated().sum()
```

```python
info_df = pd.DataFrame({
    'Column': df_cut.columns,
    "Duplicate rows": df_cut.duplicated().sum(),
    "Duplicates": [df_cut[col].duplicated().sum() for col in df_cut.columns],
    'Non-null': df_cut.notnull().sum().values,
    'Missing values': df_cut.isna().sum().values,
    'Percentage of missing values': (df_cut.isna().sum() * 100 / df_cut.notnull().sum()).values,
    'Unique values': df_cut.nunique().values,
    'Data type': df_cut.dtypes.values,
})

# Display the table
info_df
```

```python
df_cut[df_cut["id"].duplicated(keep=False)].sort_values("id").head(2)
```

## 0.3.1 Remove records with identical "id"

```python
df_cut = df_cut.drop_duplicates(subset="id").reset_index(drop=True)
```

```python
df_cut["id"].duplicated().sum()
```

```python
info_df
```

## 0.4 Remove Missing values

```python
df_cut["description"] = df_cut["description"].fillna("No description")
df_cut["homepage"] = df_cut["homepage"].fillna("No homepage")
df_cut["language"] = df_cut["language"].fillna("Unknown")
df_cut["license.name"] = df_cut["license.name"].fillna("No license")
df_cut["license.spdx_id"] = df_cut["license.spdx_id"].fillna("No license id")
```

```python
df_cut["created_at"] = pd.to_datetime(df_cut["created_at"])

print(df_cut["created_at"].min())
print(df_cut["created_at"].max())
```

#### Repositories with missing values in the language column were excluded, as they primarily represented documentation, paper lists, and other non-code resources that were outside the scope of this analysis.

```python
df_cut = df_cut[df_cut["language"].notna()].copy()
df_cut.info()
```


## 0.5 Fixing missing values

```python
df_cut["description"] = df_cut["description"].fillna("No description")
df_cut["homepage"] = df_cut["homepage"].fillna("No homepage")
df_cut["language"] = df_cut["language"].fillna("Unknown")
df_cut["license.name"] = df_cut["license.name"].fillna("No license")
df_cut["license.spdx_id"] = df_cut["license.spdx_id"].fillna("No license")
```

```python
info_df = pd.DataFrame({
    'Column': df_cut.columns,
    "Duplicate rows": df_cut.duplicated().sum(),
    "Duplicates": [df_cut[col].duplicated().sum() for col in df_cut.columns],
    'Non-null': df_cut.notnull().sum().values,
    'Missing values': df_cut.isna().sum().values,
    'Percentage of missing values': (df_cut.isna().sum() * 100 / df_cut.notnull().sum()).values,
    'Unique values': df_cut.nunique().values,
    'Data type': df_cut.dtypes.values,
})

# Display the table
info_df
```

```python
df_cut.drop(columns="fork", inplace=True)
```
