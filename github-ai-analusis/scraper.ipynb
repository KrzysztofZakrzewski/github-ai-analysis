## Topic list

```python
AI_TOPICS = [
    "artificial-intelligence",
    "ai",
    "machine-learning",
    "deep-learning",
    "llm",
    "large-language-models",
    "generative-ai",
    "gpt",
    "chatgpt",
    "rag",
    "mcp",
    "agents",
    "agentic-ai",
    "agentic-workflow",
    "nlp",
    "natural-language-processing",
    "computer-vision",
    "reinforcement-learning",
    "pytorch",
    "tensorflow",
]
```

## Function

```python
URL = "https://api.github.com/search/repositories"

def fetch_repositories(query, page=1, per_page=100):

    params = {
        "q": query,
        "sort": "stars",
        "order": "desc",
        "per_page": per_page,
        "page": page
    }

    headers = {
        "Accept": "application/vnd.github+json",
        "Authorization": f"Bearer {TOKEN}"
    }

    for attempt in range(10):

        response = requests.get(URL, params=params, headers=headers, timeout=30)

        if response.status_code == 200:
            return response.json()

        print(
            f"Błąd {response.status_code} | "
            f"query={query} | page={page} | "
            f"próba {attempt + 1}/10"
        )

        print(response.text)
        time.sleep(5)

    return None
```

### Scrap data in loop

```python
output_file = "data/raw/github_ai_repositories2.csv"

all_items = []

for topic in AI_TOPICS:

    print(f"I'm downloading: {topic}")

    topic_items = []

    for page in range(1, 11):

        data = fetch_repositories(
            query=f"topic:{topic}",
            page=page
        )

        if data is None:
            continue

        topic_items.extend(data["items"])
        all_items.extend(data["items"])

    df_topic = pd.json_normalize(topic_items)

    print(f"{topic}: {len(df_topic.columns)} columns")
    print(df_topic.columns.tolist())
    print("-" * 80)

df = pd.json_normalize(all_items)

df.to_csv(output_file, index=False)

print(f"Total {len(df)}  records saved.")
```
